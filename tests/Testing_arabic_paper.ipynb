```python
# ============================================================
# HEATMAP AND DEPENDENCE PLOTS FOR ARABIC DATASET
# ============================================================

import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# RTL library imports
from shap_rtl.plots import heatmap, scatter
from shap_rtl import RTLSHAPExplainer

# ============================================================
# CELL 1: LOAD ARABIC DATA
# ============================================================

print("="*80)
print("📊 LOADING ARABIC DATASET")
print("="*80)

# Load Arabic data
file_path = r'C:\Users\USER\OneDrive - Higher Education Commission\Dr. shahid project\SHAP_package\L-HSAB (1).txt'

# Load dataset
try:
    df = pd.read_csv(file_path, sep='\t', header=None, names=['text', 'label'])
except:
    try:
        df = pd.read_csv(file_path, sep=',', header=None, names=['text', 'label'])
    except:
        df = pd.read_csv(file_path, sep='\s+', header=None, names=['text', 'label'])

# Clean data
df = df.dropna()
df['text'] = df['text'].astype(str).str.strip()
df['label'] = df['label'].astype(str).str.strip()

# Map labels to binary (0=Normal, 1=Abusive)
unique_labels = df['label'].unique()
label_map = {}
normal_labels = ['normal', '0', 'non-abusive', 'non abusive', 'not abusive']
abusive_labels = ['abusive', '1', 'hate', 'offensive', 'hate speech']

for label in unique_labels:
    label_lower = str(label).lower().strip()
    if label_lower in normal_labels:
        label_map[label] = 0
    elif label_lower in abusive_labels:
        label_map[label] = 1
    else:
        try:
            num_label = float(label)
            if num_label == 0:
                label_map[label] = 0
            else:
                label_map[label] = 1
        except:
            if any(word in label_lower for word in ['abuse', 'hate', 'offens', 'toxic']):
                label_map[label] = 1
            else:
                label_map[label] = 0

df['label'] = df['label'].map(label_map)
df = df.dropna(subset=['label'])
df['label'] = df['label'].astype(int)

class_names = {0: 'Normal', 1: 'Abusive'}

print(f"✅ Dataset loaded: {len(df)} samples")
for class_id, name in class_names.items():
    count = (df['label'] == class_id).sum()
    print(f"  {name}: {count} ({count/len(df)*100:.1f}%)")

# ============================================================
# CELL 2: TRAIN MODEL
# ============================================================

print("\n" + "="*80)
print("🔧 TRAINING MODEL")
print("="*80)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'], 
    test_size=0.2, 
    random_state=42,
    stratify=df['label']
)

# Vectorize with character n-grams for Arabic
vectorizer = TfidfVectorizer(
    ngram_range=(1, 3),
    analyzer='word',
    max_features=5000,
    min_df=2,
    max_df=0.95
)

X_train_vec = vectorizer.fit_transform(X_train).toarray()
X_test_vec = vectorizer.transform(X_test).toarray()

# Train model
model = LogisticRegression(
    max_iter=1000, 
    class_weight='balanced', 
    random_state=42, 
    solver='lbfgs'
).fit(X_train_vec, y_train)

print(f"✅ Model trained!")
print(f"Train Accuracy: {model.score(X_train_vec, y_train):.3f}")
print(f"Test Accuracy: {model.score(X_test_vec, y_test):.3f}")

feature_names = list(vectorizer.get_feature_names_out())

# ============================================================
# CELL 3: SETUP RTL EXPLAINER
# ============================================================

print("\n" + "="*80)
print("🔧 SETUP RTL EXPLAINER")
print("="*80)

X_background = X_train_vec[:100]
masker = shap.maskers.Independent(data=X_background, max_samples=100)
shap_explainer = shap.LinearExplainer(model, masker=masker)

explainer_rtl = RTLSHAPExplainer(
    language="Arabic",
    model=model,
    vectorizer=vectorizer,
    api_type="groq"
)

print("✅ RTL SHAP explainer ready!")

# ============================================================
# CELL 4: HEATMAP PLOT - Dataset Level
# ============================================================

print("\n" + "="*80)
print("🔥 HEATMAP PLOT - Feature Patterns Across Dataset")
print("="*80)

n_samples = 50  # Number of samples to use
max_features = 15  # Number of features to show

print(f"Using {n_samples} samples from dataset")
print(f"Showing top {max_features} features")
print("-"*60)

# Get SHAP values for multiple samples
shap_values_list = []
valid_samples = []
valid_labels = []

sample_indices = np.random.choice(len(X_test_vec), min(n_samples, len(X_test_vec)), replace=False)

print("📊 Computing SHAP values...")
for idx in sample_indices:
    vec_sample = X_test_vec[idx]
    if np.sum(vec_sample) > 0:
        shap_val = shap_explainer(vec_sample.reshape(1, -1))
        shap_values_list.append(shap_val.values[0])
        valid_samples.append(vec_sample)
        valid_labels.append(y_test.iloc[idx] if hasattr(y_test, 'iloc') else y_test[idx])

if shap_values_list:
    shap_values_all = np.array(shap_values_list)
    valid_samples = np.array(valid_samples)
    valid_labels = np.array(valid_labels)
    
    # Get top features across all samples
    top_features_idx = np.argsort(np.abs(shap_values_all).mean(axis=0))[::-1][:max_features]
    top_feature_names = [feature_names[i] for i in top_features_idx]
    
    print(f"\n📊 Top {max_features} features:")
    for i, name in enumerate(top_feature_names):
        avg_impact = np.abs(shap_values_all[:, i]).mean()
        direction = "→ ABUSIVE" if np.mean(shap_values_all[:, i]) > 0 else "→ NORMAL"
        print(f"  {i+1}. '{name}': Avg impact {avg_impact:.4f} {direction}")
    
    # Create SHAP Explanation for heatmap
    shap_exp_heatmap = shap.Explanation(
        values=shap_values_all[:, top_features_idx],
        base_values=shap_explainer.expected_value if hasattr(shap_explainer, 'expected_value') else 0,
        data=valid_samples[:, top_features_idx],
        feature_names=top_feature_names
    )
    
    # Generate RTL Heatmap Plot
    print("\n🎨 Generating RTL Heatmap Plot...")
    print("💡 Each row = one sample, each column = one feature")
    print("💡 Red = High SHAP (pushes toward Abusive)")
    print("💡 Blue = Low SHAP (pushes toward Normal)")
    
    explainer_rtl.run_full_analysis(
        shap_values=shap_exp_heatmap,
        prediction_label=class_names[1],
        plot_fn=heatmap,
        max_features=max_features
    )
    plt.show()
    
    # Print interpretation
    print("\n📊 HEATMAP INTERPRETATION:")
    print("-" * 60)
    
    # Find most consistent features
    positive_ratio = np.mean(shap_values_all[:, top_features_idx] > 0, axis=0)
    for i, name in enumerate(top_feature_names):
        if positive_ratio[i] > 0.7:
            print(f"  ✅ '{name}': Consistent indicator of ABUSIVE ({positive_ratio[i]*100:.0f}% of samples)")
        elif positive_ratio[i] < 0.3:
            print(f"  ❌ '{name}': Consistent indicator of NORMAL ({(1-positive_ratio[i])*100:.0f}% of samples)")
        else:
            print(f"  ⚖️ '{name}': Mixed indicator ({positive_ratio[i]*100:.0f}% abusive)")
    
else:
    print("❌ Not enough samples for heatmap plot")

# ============================================================
# CELL 5: DEPENDENCE PLOT - Feature Interaction
# ============================================================

print("\n" + "="*80)
print("📊 DEPENDENCE PLOT - Feature Interaction")
print("="*80)

n_samples = 100  # Use more samples for dependence plot

print(f"Using {n_samples} samples from dataset")
print("-"*60)

# Get more SHAP values for dependence plot
shap_values_full = []
data_full = []
labels_full = []

sample_indices_full = np.random.choice(len(X_test_vec), min(n_samples, len(X_test_vec)), replace=False)

print("📊 Computing SHAP values...")
for idx in sample_indices_full:
    vec_sample = X_test_vec[idx]
    if np.sum(vec_sample) > 0:
        shap_val = shap_explainer(vec_sample.reshape(1, -1))
        shap_values_full.append(shap_val.values[0])
        data_full.append(vec_sample)
        labels_full.append(y_test.iloc[idx] if hasattr(y_test, 'iloc') else y_test[idx])

if shap_values_full:
    shap_values_full = np.array(shap_values_full)
    data_full = np.array(data_full)
    labels_full = np.array(labels_full)
    
    # Get top features
    top_features_full = np.argsort(np.abs(shap_values_full).mean(axis=0))[::-1][:10]
    top_feature_names_full = [feature_names[i] for i in top_features_full]
    
    print(f"\n📊 Available features for dependence plot:")
    for i, name in enumerate(top_feature_names_full[:5]):
        print(f"  {i+1}. '{name}'")
    
    # Let user choose which features to plot
    # Default: Use top 2 features
    feature_idx1 = top_features_full[0]
    feature_idx2 = top_features_full[1]
    feature_name1 = top_feature_names_full[0]
    feature_name2 = top_feature_names_full[1]
    
    print(f"\n📊 Plotting dependence for:")
    print(f"  Main feature: '{feature_name1}'")
    print(f"  Color by: '{feature_name2}'")
    print("-"*60)
    
    # Create dependence plot using matplotlib
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # Scatter plot with color mapping
    scatter = ax.scatter(
        data_full[:, feature_idx1],
        shap_values_full[:, feature_idx1],
        c=data_full[:, feature_idx2],
        cmap='RdYlBu',
        alpha=0.6,
        s=60,
        edgecolors='black',
        linewidth=0.5
    )
    
    # Add labels and title
    ax.set_xlabel(f'Feature value: {feature_name1}', fontsize=12)
    ax.set_ylabel(f'SHAP value for {feature_name1}', fontsize=12)
    ax.set_title(f'Dependence Plot: {feature_name1}\nColored by {feature_name2}', 
                 fontsize=14, fontweight='bold')
    
    # Add colorbar
    cbar = plt.colorbar(scatter)
    cbar.set_label(f'Feature value: {feature_name2}', fontsize=11)
    
    # Add horizontal line at y=0
    ax.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
    
    # Add grid
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # ============================================================
    # DEPENDENCE PLOT 2: Second Feature Pair
    # ============================================================
    
    if len(top_features_full) >= 4:
        print("\n📊 Second Dependence Plot:")
        print("-"*60)
        
        feature_idx3 = top_features_full[2]
        feature_idx4 = top_features_full[3]
        feature_name3 = top_feature_names_full[2]
        feature_name4 = top_feature_names_full[3]
        
        print(f"  Main feature: '{feature_name3}'")
        print(f"  Color by: '{feature_name4}'")
        
        fig, ax = plt.subplots(figsize=(12, 8))
        
        scatter = ax.scatter(
            data_full[:, feature_idx3],
            shap_values_full[:, feature_idx3],
            c=data_full[:, feature_idx4],
            cmap='RdYlBu',
            alpha=0.6,
            s=60,
            edgecolors='black',
            linewidth=0.5
        )
        
        ax.set_xlabel(f'Feature value: {feature_name3}', fontsize=12)
        ax.set_ylabel(f'SHAP value for {feature_name3}', fontsize=12)
        ax.set_title(f'Dependence Plot: {feature_name3}\nColored by {feature_name4}', 
                     fontsize=14, fontweight='bold')
        
        cbar = plt.colorbar(scatter)
        cbar.set_label(f'Feature value: {feature_name4}', fontsize=11)
        
        ax.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
        ax.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    # ============================================================
    # DEPENDENCE PLOT: By Sample Class
    # ============================================================
    
    print("\n📊 Dependence Plot - Colored by Sample Class:")
    print("-"*60)
    
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # Separate by class
    for class_id in [0, 1]:
        class_mask = labels_full == class_id
        if np.sum(class_mask) > 0:
            ax.scatter(
                data_full[class_mask, feature_idx1],
                shap_values_full[class_mask, feature_idx1],
                label=class_names[class_id],
                alpha=0.6,
                s=50,
                c='red' if class_id == 1 else 'blue'
            )
    
    ax.set_xlabel(f'Feature value: {feature_name1}', fontsize=12)
    ax.set_ylabel(f'SHAP value for {feature_name1}', fontsize=12)
    ax.set_title(f'Dependence Plot: {feature_name1}\nColored by Actual Class', 
                 fontsize=14, fontweight='bold')
    ax.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
    ax.grid(True, alpha=0.3)
    ax.legend()
    
    plt.tight_layout()
    plt.show()
    
    # ============================================================
    # INTERPRETATION
    # ============================================================
    
    print("\n📊 DEPENDENCE PLOT INTERPRETATION:")
    print("="*60)
    
    # Analyze the main feature
    shap_vals_main = shap_values_full[:, feature_idx1]
    data_vals_main = data_full[:, feature_idx1]
    
    # Correlation between feature value and SHAP
    correlation = np.corrcoef(data_vals_main, shap_vals_main)[0, 1]
    
    print(f"\n🔍 Feature: '{feature_name1}'")
    print(f"  • Correlation with SHAP: {correlation:.3f}")
    print(f"  • Mean SHAP value: {np.mean(shap_vals_main):.3f}")
    print(f"  • Positive SHAP: {np.mean(shap_vals_main > 0)*100:.1f}% of samples")
    
    if correlation > 0.3:
        print(f"  ✅ Strong positive correlation: Higher feature value → Higher SHAP (pushes toward Abusive)")
    elif correlation < -0.3:
        print(f"  ✅ Strong negative correlation: Higher feature value → Lower SHAP (pushes toward Normal)")
    else:
        print(f"  ⚖️ Weak correlation: Feature doesn't clearly predict direction")
    
    # Check if feature is more important for abusive or normal
    abusive_shap = shap_vals_main[labels_full == 1] if np.sum(labels_full == 1) > 0 else []
    normal_shap = shap_vals_main[labels_full == 0] if np.sum(labels_full == 0) > 0 else []
    
    if len(abusive_shap) > 0 and len(normal_shap) > 0:
        print(f"  • Avg SHAP in Abusive samples: {np.mean(abusive_shap):.3f}")
        print(f"  • Avg SHAP in Normal samples: {np.mean(normal_shap):.3f}")
        
        if np.mean(abusive_shap) > np.mean(normal_shap):
            print(f"  ✅ Feature is more POSITIVE in Abusive samples")
        else:
            print(f"  ✅ Feature is more NEGATIVE in Normal samples")
    
else:
    print("❌ Not enough samples for dependence plot")

print("\n" + "="*80)
print("✅ ANALYSIS COMPLETE!")
print("="*80)
```

```python
# ============================================================
# HEBREW DATA PREPROCESSING - Remove Usernames, Hashtags, @
# ============================================================

import re
import pandas as pd
import numpy as np

def preprocess_hebrew_text(text):
    """
    Preprocess Hebrew text by removing:
    - @mentions (usernames)
    - #hashtags
    - URLs
    - Extra whitespace
    - Special characters (optional)
    - Numbers (optional)
    
    Parameters
    ----------
    text : str
        Input text to preprocess
    
    Returns
    -------
    str
        Cleaned text
    """
    if not isinstance(text, str):
        return ""
    
    # Convert to string
    text = str(text)
    
    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    
    # Remove @mentions (usernames) - handles both English and Hebrew usernames
    text = re.sub(r'@\w+', '', text)  # English usernames
    text = re.sub(r'@[\u0590-\u05FF\w]+', '', text)  # Hebrew usernames
    
    # Remove #hashtags (keeps the text after #)
    text = re.sub(r'#\w+', '', text)  # English hashtags
    text = re.sub(r'#[\u0590-\u05FF\w]+', '', text)  # Hebrew hashtags
    
    # Remove RT (retweet indicators)
    text = re.sub(r'\bRT\b', '', text)
    
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text


def preprocess_hebrew_dataset(df, text_column, remove_numbers=True, remove_special_chars=True):
    """
    Preprocess the entire Hebrew dataset.
    
    Parameters
    ----------
    df : pandas DataFrame
        Input DataFrame
    text_column : str
        Name of the column containing text
    remove_numbers : bool
        Whether to remove numbers
    remove_special_chars : bool
        Whether to remove special characters
    
    Returns
    -------
    pandas DataFrame
        DataFrame with preprocessed text
    """
    print("\n" + "="*80)
    print("🔧 PREPROCESSING HEBREW DATASET")
    print("="*80)
    
    # Make a copy to avoid modifying original
    df_clean = df.copy()
    
    # Get original sample
    sample_text = df_clean[text_column].iloc[0] if len(df_clean) > 0 else ""
    print(f"\n📝 Original sample:")
    print(f"   {sample_text[:100]}...")
    
    # Apply preprocessing
    print(f"\n🔄 Applying preprocessing...")
    df_clean['text_clean'] = df_clean[text_column].apply(preprocess_hebrew_text)
    
    # Optional: Remove numbers
    if remove_numbers:
        print("   - Removing numbers...")
        df_clean['text_clean'] = df_clean['text_clean'].apply(
            lambda x: re.sub(r'\d+', '', x) if isinstance(x, str) else x
        )
    
    # Optional: Remove special characters (keep Hebrew and spaces)
    if remove_special_chars:
        print("   - Removing special characters...")
        # Keep Hebrew characters, spaces, and basic punctuation
        df_clean['text_clean'] = df_clean['text_clean'].apply(
            lambda x: re.sub(r'[^\u0590-\u05FF\s]', '', x) if isinstance(x, str) else x
        )
    
    # Remove extra whitespace again
    df_clean['text_clean'] = df_clean['text_clean'].apply(
        lambda x: re.sub(r'\s+', ' ', str(x)).strip() if isinstance(x, str) else x
    )
    
    # Show results
    cleaned_sample = df_clean['text_clean'].iloc[0] if len(df_clean) > 0 else ""
    print(f"\n✅ Cleaned sample:")
    print(f"   {cleaned_sample[:100]}...")
    
    # Statistics
    original_length = df_clean[text_column].astype(str).str.len().mean()
    cleaned_length = df_clean['text_clean'].astype(str).str.len().mean()
    print(f"\n📊 Statistics:")
    print(f"   Average original length: {original_length:.1f} characters")
    print(f"   Average cleaned length: {cleaned_length:.1f} characters")
    print(f"   Reduction: {(1 - cleaned_length/original_length)*100:.1f}%")
    
    return df_clean


# ============================================================
# COMPLETE PREPROCESSING PIPELINE
# ============================================================

def complete_preprocessing_pipeline(file_path, text_column, label_column):
    """
    Complete preprocessing pipeline for Hebrew dataset.
    
    Parameters
    ----------
    file_path : str
        Path to the CSV file
    text_column : str
        Name of the text column
    label_column : str
        Name of the label column
    
    Returns
    -------
    pandas DataFrame
        Fully preprocessed DataFrame
    """
    print("\n" + "="*80)
    print("📊 COMPLETE PREPROCESSING PIPELINE")
    print("="*80)
    
    # 1. Load data
    print(f"\n📂 Loading data from: {file_path}")
    df = pd.read_csv(file_path)
    print(f"   Loaded {len(df)} samples")
    
    # 2. Clean labels
    print(f"\n🏷️ Cleaning labels...")
    
    def map_hebrew_label(label):
        label_lower = str(label).lower().strip()
        violence_keywords = ['violence', 'violent', 'hate', 'abusive', 'hate,', 'violence,', 'racism', '1']
        for keyword in violence_keywords:
            if keyword in label_lower:
                return 1
        if label_lower in ['not', '0']:
            return 0
        try:
            num = float(label)
            return 1 if num == 1 else 0
        except:
            return 0
    
    df['label'] = df[label_column].apply(map_hebrew_label)
    df['label'] = df['label'].astype(int)
    
    # 3. Preprocess text
    df = preprocess_hebrew_dataset(df, text_column)
    
    # 4. Remove empty texts
    df = df[df['text_clean'].str.len() > 0]
    
    print(f"\n✅ Final dataset: {len(df)} samples")
    
    # Show class distribution
    class_names = {0: 'NOT Violence', 1: 'VIOLENCE'}
    print(f"\n📊 Class Distribution:")
    for class_id, name in class_names.items():
        count = (df['label'] == class_id).sum()
        print(f"  {name}: {count} ({count/len(df)*100:.1f}%)")
    
    return df


# ============================================================
# USAGE EXAMPLE
# ============================================================

if __name__ == "__main__":
    # File path
    file_path = r'C:\Users\USER\OneDrive - Higher Education Commission\Dr. shahid project\SHAP_package\AllData_OffensiveHebrew.csv'
    
    # Run preprocessing
    df_clean = complete_preprocessing_pipeline(
        file_path=file_path,
        text_column='TweetText',
        label_column='Label'
    )
    
    # Display sample
    print(f"\n📝 Sample cleaned texts:")
    print("-"*60)
    for i in range(min(5, len(df_clean))):
        print(f"{i+1}. {df_clean['text_clean'].iloc[i][:80]}...")
    
    # Save cleaned data
    output_path = file_path.replace('.csv', 'hebrew_cleaned.csv')
    df_clean.to_csv(output_path, index=False)
    print(f"\n💾 Cleaned data saved to: {output_path}")
```
